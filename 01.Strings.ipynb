```python
import re
from dataclasses import dataclass
from typing import Any, Iterable


UNSAFE_CHARS_PATTERN = re.compile(r"[^a-z0-9]+")
MULTIPLE_UNDERSCORES_PATTERN = re.compile(r"_+")
VALID_IDENTIFIER_PATTERN = re.compile(r"^[a-z_][a-z0-9_]*$")


@dataclass(frozen=True, slots=True)
class ColumnNormalizationResult:
    original_column_name: Any
    normalized_column_name: str | None
    valid: bool
    collision: bool
    error: str | None = None


def normalize_identifier(column: Any) -> str:
    """
    Normalize a single source-system column name into a
    warehouse-safe identifier.

    Examples:
        " First Name " -> "first_name"
        "USER__ID__"   -> "user_id"
        "Total ($)"    -> "total"
    """

    if not isinstance(column, str):
        raise TypeError(
            f"Column name must be a string, got {type(column).__name__}"
        )

    value = column.strip()

    if not value:
        raise ValueError("Column name cannot be blank")

    # Normalize casing.
    value = value.lower()

    # Convert punctuation, whitespace, and other unsafe runs
    # into a single underscore.
    value = UNSAFE_CHARS_PATTERN.sub("_", value)

    # Defensive collapse in case underscores already existed.
    value = MULTIPLE_UNDERSCORES_PATTERN.sub("_", value)

    # Remove separators from both ends.
    value = value.strip("_")

    if not value:
        raise ValueError(
            f"Column name {column!r} contains no usable characters"
        )

    # SQL-style identifiers should generally not start with a digit.
    if value[0].isdigit():
        value = f"col_{value}"

    if not VALID_IDENTIFIER_PATTERN.fullmatch(value):
        raise ValueError(
            f"Unable to create valid identifier from {column!r}"
        )

    return value


def normalize_column_names(
    columns: Iterable[Any],
) -> list[ColumnNormalizationResult]:
    """
    Normalize a collection of columns and deterministically
    resolve collisions with numeric suffixes.

    Example:
        ["first-name", "first_name", "First Name"]

    becomes:
        ["first_name", "first_name_2", "first_name_3"]
    """

    results: list[ColumnNormalizationResult] = []

    # Tracks every final output name.
    used_names: set[str] = set()

    # Tracks the next suffix for each base normalized name.
    counters: dict[str, int] = {}

    for column in columns:
        try:
            base_name = normalize_identifier(column)

        except (TypeError, ValueError) as exc:
            results.append(
                ColumnNormalizationResult(
                    original_column_name=column,
                    normalized_column_name=None,
                    valid=False,
                    collision=False,
                    error=str(exc),
                )
            )
            continue

        normalized_name = base_name
        collision = normalized_name in used_names

        if collision:
            next_suffix = counters.get(base_name, 2)

            while f"{base_name}_{next_suffix}" in used_names:
                next_suffix += 1

            normalized_name = f"{base_name}_{next_suffix}"
            counters[base_name] = next_suffix + 1

        else:
            counters.setdefault(base_name, 2)

        used_names.add(normalized_name)

        results.append(
            ColumnNormalizationResult(
                original_column_name=column,
                normalized_column_name=normalized_name,
                valid=True,
                collision=collision,
            )
        )

    return results


if __name__ == "__main__":
    column_list = [
        "   first_name  ",
        "first-name",
        "total  ($)",
        "USER__ID__",
        " ",
        "first_name",
        "#_of_Items",
    ]

    results = normalize_column_names(column_list)

    for result in results:
        print(result)
```

```python
#Validate Composite Business Keys-Validate business-key strings composed of multiple components such as COUNTRY|ACCOUNT_ID|YYYYMMDD. Reject missing components, unexpected delimiters, whitespace-only values, and invalid component lengths.
import re
from datetime import datetime
from typing import Any, Optional
from dataclasses import dataclass

VALID_DELIMITER='|'
VALID_KEY_LENGTH=3

#regex patterns for individual key components
COUNTRY_CODE_PATTERN=re.compile(r"[A-Z]{3}")
ACCOUNT_ID_PATTERN=re.compile(r"ACC\d{6}")
DATE_PATTERN=re.compile(r"\d{8}")
DELIMITER_PATTERN=re.compile(r"[-_,;:/\\]")

#result class
@dataclass(frozen=True, slots=True)
class ParsingKeyResults:
    original_key:Optional[str]=None
    country_code:Optional[str]=None
    account_id:Optional[str]=None
    date_part:Optional[date]=None
    delimiter:Optional[str]=None
    is_valid:bool=False
    error_code:Optional[str]=None
    error:Optional[str]=None

#errors in separate method
def _invalid(error:str, error_code:str)->ParsingKeyResults:
    return ParsingKeyResults(
        is_valid=False,
        error_code=error_code,
        error=error
    )

#method
def key_parsing(key:object)->ParsingKeyResults:
    #check if given key is string
    if not isinstance(key, str):
        return _invalid(
            f"Invalid_type",
            f"type of key expected to be string but got {type(key).__name__}"
        )
    #check if given key is blank
    normalized_key=key.strip()
    if not normalized_key:
        return _invalid(
            f"Empty key",
            f"Key cannot be empty/white spaces"
        )
    #check type of delimiters
    unexpected_delimiter=DELIMITER_PATTERN.search(normalized_key)
    if unexpected_delimiter:
        return _invalid(
            f"invalid delimiter found {unexpected_delimiter.group()}",
            f"expected delmiter is {VALID_DELIMITER}"
        )
    #check count of delimiters
    delimiter_count=normalized_key.count(VALID_DELIMITER)
    if delimiter_count!=(VALID_KEY_LENGTH-1):
        return _invalid(
            f"business key must contain {VALID_KEY_LENGTH-1} delimiters",
            f"received {delimiter_count} delimiters"
        ) 
    parts=normalized_key.split(VALID_DELIMITER)
    #check length of key componennts
    if len(parts)!=VALID_KEY_LENGTH:
        return _invalid(
            f"INVALID STRUCTURE",
            f"key must have {VALID_KEY_LENGTH} components received {len(parts)}"
        )     
    #check if one or more parts are white spaces
    if any(not part.strip() for part in parts):
        return _invalid(
            f"Missing component",
            f"business key contain empty or white spaced components"
        )
    if any(part!=part.strip() for part in parts):
        return _invalid(
            f"components not of standard length",
            f"business key contain one or more trailing or heading white spaces"
        )

    country_code, account_id, date_part=parts
    #check country code     
    if not COUNTRY_CODE_PATTERN.fullmatch(country_code):
        return _invalid(
            f"invalid country code",
            f"country code must contain 3 uppercase english letters"
        )
    #check account number
    if not ACCOUNT_ID_PATTERN.fullmatch(account_id):
        return _invalid(
            f"invalid account id",
            f"account number must begin with ACC forllowed by 6 digits"
        ) 
    #check date part is valid
    if not DATE_PATTERN.fullmatch(date_part):
        return _invalid(
            f"invlid date",
            f"date must contain 8 digits with format YYYYMMDD"
        )
    #check if date matches with calendar date
    try:
        business_date=datetime.strptime(date_part,'%Y%m%d').date()
    except ValueError:
        return _invalid(
            f"invalid date",
            f"this is not valid calendar date"
        )
    #return results
    return ParsingKeyResults(
            original_key=normalized_key,
            country_code=country_code,
            account_id=account_id,
            date_part=date_part,
            delimiter=VALID_DELIMITER,
            is_valid=True
    )

if __name__=="__main__":
    key_list = [
        "USA-ACC849204-20260818",       # Invalid delimiter
        "CAN|ACC102943|20251231",       # Valid
        "GBR|ACC554121|20260115",       # Valid
        "DEU|ACC003921|20261318",       # Invalid month
        "AUS|ACC773910|20260632",       # Invalid day
        "JPN||20260519",                # Missing component
        "FRA|ACC883012|20220910|",      # Trailing delimiter
        "SGP|ACC229481|20250430",       # Valid
        "BRA|ACC665039|20260228",       # Valid
        "INDACC33492020211015",         # Missing delimiters
        "MEX|ACC112233|20240229",       # Valid leap-day date
        "USA|ACC1|20260818",            # Invalid account length
        "GBR|ACC99X7Z|202603122",       # Invalid account/date
        "CAN|ACC102943|20351231",       # Valid unless future dates are forbidden
        "USA|   |20260818",             # Whitespace-only component
        "USA |ACC123456|20260818",      # Component whitespace
    ]

    for key in key_list:
        result=key_parsing(key)
        print(f"{key!r}:{result}")
```
