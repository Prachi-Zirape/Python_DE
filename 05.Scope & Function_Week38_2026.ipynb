```python
#Standardize Date Fields — Create a function that accepts dates arriving in multiple formats and returns a standardized date value while safely handling invalid dates, nulls, impossible calendar values, and unsupported formats.

from datetime import datetime
from typing import Final

DEFAULT_DATE:Final[str]='9999/12/31'
DEFAULT_DATE_FORMAT:Final[str]='%Y/%m/%d'

#order preservation matters when accepted formats overlaps
ACCEPTED_DATE_FORMAT:Final[tuple[str,...]]=(
    "%m/%d/%Y",
    "%m-%d-%Y",
    "%Y-%m-%d",
    "%Y/%m/%d",
    "%B %d, %Y",
    "%Y-%m-%dT%H:%M:%SZ",
    "%Y-%m-%d %H:%M:%S",
)

def standardize_date_formats(inbound_date:str|None)->str:
    """
    Convert a supported inbound date string into YYYY/MM/DD.

    Missing values (None, empty, or whitespace-only strings) return the
    business-approved default date. Invalid or unsupported values raise
    ValueError so the calling pipeline can reject or quarantine the record.

    Args:
        inbound_date: A date string in one of ACCEPTED_DATE_FORMATS, or None.

    Returns:
        A standardized date string in YYYY/MM/DD format.

    Raises:
        TypeError: If inbound_date is neither a string nor None.
        ValueError: If the value is non-empty but invalid or unsupported.
    """
    if inbound_date is None:
        return DEFAULT_DATE
    if not isinstance(inbound_date, str):
        raise TypeError(f"date expected to be string or None"
                        f"received {type(inbound_date).__name__}")
    cleaned_date=inbound_date.strip()
    if not cleaned_date:
        return DEFAULT_DATE
    
    for _format in ACCEPTED_DATE_FORMAT:
        try:
            parsed_date=datetime.strptime(cleaned_date,_format ).date()
            return parsed_date.strftime(DEFAULT_DATE_FORMAT)
        except ValueError:
            continue
    supported_formats= ", ".join(ACCEPTED_DATE_FORMAT)
    raise ValueError(
        f"unsupported or invalid values {cleaned_date!r}"
        f"supported formats {supported_formats}")
    
if __name__=="__main__":
    inbound_dates=[
        "2026-09-14",
        "09/14/2026",
        "2026/09/14",
        "2026-01-00",
        "2026-09-14T10:30:00Z",
        "abc2026-09-14",
        "not a date",
        "",
        None,
    ]
    for inbound_date in inbound_dates:
        try:
            result=standardize_date_formats(inbound_date)
            print(f"{inbound_date!r}->{result}")
        except (TypeError, ValueError) as exc:
            print(f"{inbound_date!r}->rejected {exc}")
```
